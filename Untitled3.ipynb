```python
tropics = ds_gpcp["precip"].sel(lat=slice(-20, 20))
data1 = tropics.mean(dim=("lat", "lon"))  # 1D (time)

plt.figure(figsize=(8, 3))
data1.plot()
plt.title("Tropical-mean GPCP precipitation (20°S–20°N)")
plt.tight_layout()
data2 = tropics.mean(dim="time")         # 2D (lat, lon)

plt.figure(figsize=(8, 2.8))
data2.plot(cmap="rainbow", add_colorbar=True)
plt.title("Time-mean GPCP precipitation (1979–2021, 20°S–20°N)")
plt.tight_layout()
```
